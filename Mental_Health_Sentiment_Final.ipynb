```python
df["text_clean"]= df["post_text"].copy()
```

```python
import re
def clean_text(text):
    text = str(text).lower()
    text = text.replace("'","'")
    text = re.sub(r'\brt\b','',text)
    text = re.sub(r'http\S+|www\S+','',text)
    text = re.sub(r'@\w+','',text)
    text = text.replace('#','')
    text = re.sub(r"[^\w\s']",' ',text)
    text = re.sub(r'\s+',' ',text)
    text = text.strip()
    return text 
```

```python
df["text_clean"] = df["post_text"].apply(clean_text)
```

```python
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer
analyzer = SentimentIntensityAnalyzer()
```

```python
text = df["post_text"].iloc[0]
scores = analyzer.polarity_scores(text)
print(text)
print(scores)
```

```python
df["sentiment_score"] = df["post_text"].apply(
    lambda x: analyzer.polarity_scores(x)["compound"]
)
def sentiment_category(score):
    if score >= 0.05:
        return "Positive"
    elif score <= -0.05:
        return "Negative"
    else:
        return "Neutral"
df["sentiment"] = df["sentiment_score"].apply(sentiment_category)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tfidf = TfidfVectorizer(
    max_features=5000,
    min_df=5,
    max_df=0.95,
    stop_words="english"
)
```

```python
X_tfidf = tfidf.fit_transform(df["text_clean"])
```

```python
print("TF-IDF shape:", X_tfidf.shape)
```

```python
feature_names = tfidf.get_feature_names_out()
print(feature_names[:50])
```

```python
tfidf_df = pd.DataFrame(
    X_tfidf.toarray(),
    columns=feature_names,
    index=df.index
)
label_tfidf = tfidf_df.groupby(df["label"]).mean()
term_difference = label_tfidf.loc[1] - label_tfidf.loc[0]
top_label_1 = term_difference.sort_values(ascending=False).head(20)
top_label_0 = term_difference.sort_values().head(20)

print("Top terms associated with Label 1:")
print(top_label_1)

print("\nTop terms associated with Label 0:")
print(top_label_0)
```
